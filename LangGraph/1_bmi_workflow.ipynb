```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# define state
class BMIState(TypedDict):
    weight: float  # in kilograms
    height: float  # in meters
    bmi: float     # Body Mass Index
    category: str  # BMI category
```

```python
def calculate_bmi(state: BMIState) -> BMIState:
    weight = state["weight"]
    height = state["height"]
    bmi = weight / (height ** 2)
    state["bmi"] = bmi # partial update in state
    return state
```

```python
def label_bmi(state: BMIState) -> BMIState:
    bmi = state["bmi"]
    if bmi < 18.5:
        category = "Underweight"
    elif 18.5 <= bmi < 24.9:
        category = "Normal weight"
    elif 25 <= bmi < 29.9:
        category = "Overweight"
    else:
        category = "Obesity"
    state["category"] = category # partial update in state
    return state
```

```python
# define the graph
graph = StateGraph(BMIState)

graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('label_bmi', label_bmi)

#add edges to your graph
graph.add_edge(START, 'calculate_bmi') # START -> calculate_bmi -> label_bmi -> END
graph.add_edge('calculate_bmi', 'label_bmi')
graph.add_edge('label_bmi', END)

# compile the graph
workflow = graph.compile()
```

```python
#execute the workflow
initial_state = {"weight": 70, "height": 1.75, "bmi": 0.0}
final_state = workflow.invoke(initial_state)
print(final_state)
```
